## Parse GEF-CPT files

This example shows how to parse GEF-CPT files.

Tests with files in evo-data-converters-gef directory `tests/data/cpt/*.gef`. To provide your own files, set `SEEQUENT_EVO_TESTFILE_GEF_CPT` to a file glob, eg `/path-to-samples/gef/cpt/*/*.gef`.

```python
import os

from evo.data_converters.gef.importer.parse_gef_files import parse_gef_file

file = os.path.join(f"{os.getcwd()}/../../tests/data/cpt/cpt.gef")
print(file)

try:
    cpt = parse_gef_file(file)
    print(f"Read GEF CPT file: {file}")
    # print(cpt)
except Exception as e:
    print(f"Not a GEF CPT file: {file}")
    print(f"{type(e).__name__}: {e}")
```

```python
import glob
import os

from evo.data_converters.gef.importer.parse_gef_files import parse_gef_file

pattern = os.getenv("SEEQUENT_EVO_TESTFILE_GEF_CPT", f"{os.getcwd()}/../../tests/data/cpt/*gef")
test_files = glob.glob(pattern)

for file in test_files:
    try:
        cpt = parse_gef_file(file)
        print(f"Read GEF CPT file: {file}")
    except Exception as e:
        print(f"Not a GEF CPT file: {file}")
        print(f"{type(e).__name__}: {e}")
```
